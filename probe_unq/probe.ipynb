```python
import os, json
root='/kaggle/input/models'
for r,d,f in os.walk(root):
    if 'config.json' in f:
        try:
            c=json.load(open(os.path.join(r,'config.json')))
            print('MODEL:', r, '| hidden=', c.get('hidden_size'), '| arch=', (c.get('architectures') or ['?'])[0])
        except Exception as e:
            print('MODEL:', r, '(config unreadable)')
print('TOP:', sorted(os.listdir('/kaggle/input')) if os.path.isdir('/kaggle/input') else 'none')
```
