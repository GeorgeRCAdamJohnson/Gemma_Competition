```python

import base64, glob, json, os, re, shutil, subprocess, sys, tarfile, tempfile
from pathlib import Path
LOC = base64.b64decode("IiIibG9jYXRlOiByYW5rIHRoZSByZXBvIGZpbGVzIHRoYXQgREVGSU5FIHdoYXQgdGhlIHByb2JsZW0gdGFsa3MgYWJvdXQsIHNvIHRoZSBlZGl0IGxhbmRzIGF0IHRoZSBkZWZpbml0aW9uIHNpdGUKKG5vdCBhIGNhbGxlciwgc2libGluZywgdGVzdCwgZG9jIG9yIHBhY2thZ2luZyBmaWxlKS4gUHVyZSBzdGRsaWIgKyBnaXQ7IHJ1bnMgaW5zaWRlIHRoZSB0YXNrIHNhbmRib3ggKGN3ZCA9IHJlcG8pLgoKdXNhZ2UgKHZpYSBydW5fc2tpbGxfc2NyaXB0KTogIGFyZ3MgPSB7InEiOiAiPGlkZW50aWZpZXJzLCBjbGFzcy9mdW5jdGlvbiBuYW1lcywgZXJyb3IgdGV4dCwgZmlsZSBuYW1lcyBmcm9tIHRoZSBwcm9ibGVtPiJ9CmFsc286IHB5dGhvbjMgbG9jYXRlLnB5IC0tcSAiLi4uIiAgIG9yICAgcHl0aG9uMyBsb2NhdGUucHkgInRva2VuMSB0b2tlbjIgLi4uIgpPdXRwdXQ6IHRoZSB0b3AgY2FuZGlkYXRlIGZpbGVzIHdpdGggdGhlIGRlZmluaW5nIGxpbmVzLCBiZXN0IGZpcnN0LgoiIiIKaW1wb3J0IG1hdGgKaW1wb3J0IG9zCmltcG9ydCByZQppbXBvcnQgc3VicHJvY2VzcwppbXBvcnQgc3lzCmZyb20gY29sbGVjdGlvbnMgaW1wb3J0IGRlZmF1bHRkaWN0CgpTVE9QID0gc2V0KCIiImEgYW4gYW5kIGFyZSBhcyBhdCBiZSBiZWVuIGJ1dCBieSBjYW4gY291bGQgZG8gZG9lcyBmb3IgZnJvbSBoYXMgaGF2ZSBob3cgaWYgaW4gaW50byBpcyBpdCBpdHMgbWF5IG11c3Qgbm90Cm9mIG9uIG9yIHNob3VsZCBzbyB0aGFuIHRoYXQgdGhlIHRoZWlyIHRoZW4gdGhlcmUgdGhlc2UgdGhpcyB0byB1c2UgdXNlZCB1c2VzIHVzaW5nIHdhcyB3ZXJlIHdoZW4gd2hlcmUgd2hpY2ggd2hpbGUKd2lsbCB3aXRoIHdvdWxkIHlvdSB5b3VyIGZpeCBidWcgaXNzdWUgZXJyb3IgcHJvYmxlbSBzaG91bGQgZXhwZWN0ZWQgYWN0dWFsIGJlaGF2aW9yIGJlaGF2aW91ciBleGFtcGxlIGRvY3MgZG9jCnRydWUgZmFsc2Ugbm9uZSBzZWxmIGNscyByZXR1cm4gdmFsdWUgdmFsdWVzIGRlZmF1bHQgdHlwZSB0eXBlcyBmaWxlIGZpbGVzIGZ1bmN0aW9uIGZ1bmN0aW9ucyBtZXRob2QgY2xhc3MgY29kZQp0ZXN0IHRlc3RzIHN1cHBvcnQgYWRkIGFkZGVkIHJlbW92ZSBtYWtlIG5ldyBvbGQgd29yayB3b3JrcyB3b3JraW5nIGNhc2Ugc2FtZSBkaWZmZXJlbnQgYWxzbyBvbmx5IGp1c3QgbGlrZSIiIi5zcGxpdCgpKQpMT1cgPSByZS5jb21waWxlKHIiKF58LykoZG9jcz98dGVzdHM/fHRlc3Rpbmd8ZXhhbXBsZXM/fGJlbmNobWFya3M/fHNjcmlwdHM/KS98KF58LykodGVzdF9bXi9dKnxbXi9dKl90ZXN0XC5weXxjb25mdGVzdFwucHkpJHwiCiAgICAgICAgICAgICAgICAgciJcLihtZHxyc3R8dHh0fHRvbWx8Y2ZnfGluaXx5YW1sfHltbHxqc29ufGxvY2t8Y2ZmKSR8KF58LykoTElDRU5TRXxSRUFETUV8Q0hBTkdFTE9HfENPTlRSSUJVVElOR3xzZXR1cFwucHkpIiwKICAgICAgICAgICAgICAgICByZS5JKQoKCmRlZiByZXBvX3Jvb3QoKToKICAgIGZvciBjIGluIFtwIGZvciBwIGluIG9zLmVudmlyb24uZ2V0KCJQWVRIT05QQVRIIiwgIiIpLnNwbGl0KG9zLnBhdGhzZXApIGlmIHBdICsgWyIvd29ya3NwYWNlIiwgb3MuZ2V0Y3dkKCldOgogICAgICAgIHRyeToKICAgICAgICAgICAgciA9IHN1YnByb2Nlc3MucnVuKCJnaXQgcmV2LXBhcnNlIC0tc2hvdy10b3BsZXZlbCIsIHNoZWxsPVRydWUsIGN3ZD1jLCBjYXB0dXJlX291dHB1dD1UcnVlLCB0ZXh0PVRydWUsIHRpbWVvdXQ9MTApCiAgICAgICAgICAgIGlmIHIucmV0dXJuY29kZSA9PSAwIGFuZCByLnN0ZG91dC5zdHJpcCgpOgogICAgICAgICAgICAgICAgcmV0dXJuIHIuc3Rkb3V0LnN0cmlwKCkKICAgICAgICBleGNlcHQgRXhjZXB0aW9uOgogICAgICAgICAgICBwYXNzCiAgICByZXR1cm4gb3MuZ2V0Y3dkKCkKCgpkZWYgcXVlcnlfdGV4dCgpOgogICAgYSA9IHN5cy5hcmd2WzE6XQogICAgaWYgIi0tcSIgaW4gYToKICAgICAgICBpID0gYS5pbmRleCgiLS1xIikKICAgICAgICByZXR1cm4gIiAiLmpvaW4oYVtpICsgMTpdKQogICAgcmV0dXJuICIgIi5qb2luKHggZm9yIHggaW4gYSBpZiBub3QgeC5zdGFydHN3aXRoKCItLSIpKQoKCmRlZiB0b2tlbnMocSk6CiAgICB0b2tzID0gc2V0KCkKICAgIGZvciBtIGluIHJlLmZpbmRpdGVyKHIiYChbXmBdezIsODB9KWAiLCBxKTogICAgICAgICAgICAgICAgICAgICAgIyBiYWNrdGlja2VkIHNwYW5zCiAgICAgICAgdG9rcy51cGRhdGUocmUuZmluZGFsbChyIltBLVphLXpfXVtBLVphLXowLTlfXXsyLH0iLCBtLmdyb3VwKDEpKSkKICAgIHRva3MudXBkYXRlKHJlLmZpbmRhbGwociJcYltBLVpdW2EtejAtOV0rKD86W0EtWl1bQS1aYS16MC05XSspK1xiIiwgcSkpICAgIyBDYW1lbENhc2UKICAgIHRva3MudXBkYXRlKHJlLmZpbmRhbGwociJcYlthLXpdW2EtejAtOV0qKD86X1thLXowLTldKykrXGIiLCBxKSkgICAgICAgICAjIHNuYWtlX2Nhc2UKICAgIHRva3MudXBkYXRlKHJlLmZpbmRhbGwociJcYltBLVphLXpfXVtBLVphLXowLTlfXSooPz1cKCkiLCBxKSkgICAgICAgICAgICAjIG5hbWUoCiAgICBmb3IgbSBpbiByZS5maW5kaXRlcihyIlxiW0EtWmEtel9dW1x3XSooPzpcLltBLVphLXpfXVx3KikrXGIiLCBxKTogICAgICAjIGRvdHRlZCBhLmIuYyAtPiBwYXJ0cwogICAgICAgIHRva3MudXBkYXRlKHAgZm9yIHAgaW4gbS5ncm91cCgwKS5zcGxpdCgiLiIpIGlmIGxlbihwKSA+IDIpCiAgICB0b2tzLnVwZGF0ZShyZS5maW5kYWxsKHIiXGJbQS1aYS16X11cd3szLH1cYiIsIHEpKSAgICAgICAgICAgICAgICAgICAgICAgIyBwbGFpbiB3b3JkcyAobG93IHdlaWdodCBiZWxvdykKICAgIHJldHVybiB7dCBmb3IgdCBpbiB0b2tzIGlmIHQubG93ZXIoKSBub3QgaW4gU1RPUCBhbmQgbm90IHQuaXNkaWdpdCgpfQoKCmRlZiBmaWxlc19ieV9leHQocm9vdCk6CiAgICB0cnk6CiAgICAgICAgciA9IHN1YnByb2Nlc3MucnVuKCJnaXQgbHMtZmlsZXMiLCBzaGVsbD1UcnVlLCBjd2Q9cm9vdCwgY2FwdHVyZV9vdXRwdXQ9VHJ1ZSwgdGV4dD1UcnVlLCB0aW1lb3V0PTYwKQogICAgICAgIGZzID0gW2YgZm9yIGYgaW4gci5zdGRvdXQuc3BsaXRsaW5lcygpIGlmIGYuZW5kc3dpdGgoIi5weSIpXQogICAgZXhjZXB0IEV4Y2VwdGlvbjoKICAgICAgICBmcyA9IFtdCiAgICBpZiBub3QgZnM6ICAjIG5vIGdpdCBpbmRleCAob3IgZ2l0IG1pc3NpbmcpOiB3YWxrIHRoZSB0cmVlCiAgICAgICAgZm9yIGQsIGRuLCBmbiBpbiBvcy53YWxrKHJvb3QpOgogICAgICAgICAgICBkbls6XSA9IFt4IGZvciB4IGluIGRuIGlmIG5vdCB4LnN0YXJ0c3dpdGgoIi4iKSBhbmQgeCBub3QgaW4gKCJub2RlX21vZHVsZXMiLCAiX19weWNhY2hlX18iLCAidmVudiIsICIudmVudiIpXQogICAgICAgICAgICBmcyArPSBbb3MucGF0aC5yZWxwYXRoKG9zLnBhdGguam9pbihkLCB4KSwgcm9vdCkgZm9yIHggaW4gZm4gaWYgeC5lbmRzd2l0aCgiLnB5IildCiAgICByZXR1cm4gZnMKCgpkZWYgbWFpbigpOgogICAgcm9vdCA9IHJlcG9fcm9vdCgpCiAgICBxID0gcXVlcnlfdGV4dCgpCiAgICBpZiBub3QgcS5zdHJpcCgpOgogICAgICAgIHByaW50KCJsb2NhdGU6IHBhc3MgdGhlIHByb2JsZW0ncyBpZGVudGlmaWVycyAvIGVycm9yIHRleHQsIGUuZy4gYXJncyB7XCJxXCI6IFwiRmlsZVByb3h5IHdyaXRlIGZsdXNoXCJ9IikKICAgICAgICByZXR1cm4KICAgIHRva3MgPSB0b2tlbnMocSkKICAgIHN0cm9uZyA9IHt0IGZvciB0IGluIHRva3MgaWYgKCJfIiBpbiB0IG9yIHJlLnNlYXJjaChyIlthLXpdW0EtWl0iLCB0KSBvciByZS5zZWFyY2gocmYiYFteYF0qXGJ7cmUuZXNjYXBlKHQpfVxiIiwgcSkpfQogICAgcHlmaWxlcyA9IGZpbGVzX2J5X2V4dChyb290KQogICAgdGV4dHMgPSB7fQogICAgZm9yIGYgaW4gcHlmaWxlczoKICAgICAgICB0cnk6CiAgICAgICAgICAgIHRleHRzW2ZdID0gb3Blbihvcy5wYXRoLmpvaW4ocm9vdCwgZiksIGVuY29kaW5nPSJ1dGYtOCIsIGVycm9ycz0icmVwbGFjZSIpLnJlYWQoKQogICAgICAgIGV4Y2VwdCBPU0Vycm9yOgogICAgICAgICAgICBwYXNzCiAgICBuID0gbWF4KDEsIGxlbih0ZXh0cykpCiAgICAjIHYyICgyMDI2LTExLTEyKTogUkFSSVRZIHdlaWdodGluZy4gdjEgcmFua2VkIGh1YiBmaWxlcyAoYXBwbGljYXRpb25zLnB5LCByb3V0aW5nLnB5LCBjb25zb2xlLnB5KSBmaXJzdCBiZWNhdXNlCiAgICAjIHRoZXkgbWVudGlvbiBldmVyeXRoaW5nIC0tIHRoZSBzYW1lIHRyYXAgdGhlIGZhaWxpbmcgYWdlbnQgcnVucyBmYWxsIGludG8uIEEgdG9rZW4gaW4gMiBmaWxlcyBub3cgY291bnRzIGZhciBtb3JlCiAgICAjIHRoYW4gb25lIGluIDQwLCBmb3IgZGVmaW5pdGlvbnMgYW5kIG1lbnRpb25zIGFsaWtlLgogICAgZGYgPSB7dDogc3VtKDEgZm9yIHggaW4gdGV4dHMudmFsdWVzKCkgaWYgdCBpbiB4KSBmb3IgdCBpbiB0b2tzfQogICAgaWRmID0ge3Q6IG1hdGgubG9nKChuICsgMSkgLyAoZGZbdF0gKyAxKSkgKyAwLjEgZm9yIHQgaW4gdG9rc30KICAgIGtleSA9IGxhbWJkYSBzOiByZS5zdWIociJbXmEtejAtOV0iLCAiIiwgcy5sb3dlcigpKSAgICMgRmlsZVByb3h5ID09IGZpbGVfcHJveHkgPT0gX2ZpbGVfcHJveHkKICAgIHNjb3JlLCBkZWZzLCB1c2VzID0gZGVmYXVsdGRpY3QoZmxvYXQpLCBkZWZhdWx0ZGljdChsaXN0KSwgZGVmYXVsdGRpY3QoaW50KQogICAgZGVmX3JlID0ge3Q6IHJlLmNvbXBpbGUocmYiXlxzKig/OmFzeW5jXHMrKT8oPzpkZWZ8Y2xhc3MpXHMre3JlLmVzY2FwZSh0KX1cYnxeXHMqe3JlLmVzY2FwZSh0KX1ccypbOj1dIikgZm9yIHQgaW4gdG9rc30KICAgIGZvciBmLCB0ZXh0IGluIHRleHRzLml0ZW1zKCk6CiAgICAgICAgbG93ID0gMC4yNSBpZiBMT1cuc2VhcmNoKGYpIGVsc2UgMS4wCiAgICAgICAgc3RlbSA9IGtleShvcy5wYXRoLnNwbGl0ZXh0KG9zLnBhdGguYmFzZW5hbWUoZikpWzBdKQogICAgICAgIGZvciB0IGluIHRva3M6CiAgICAgICAgICAgIHN0ZW1faGl0ID0ga2V5KHQpID09IHN0ZW0gYW5kIGxlbihzdGVtKSA+IDMKICAgICAgICAgICAgaWYgdCBub3QgaW4gdGV4dCBhbmQgbm90IHN0ZW1faGl0OgogICAgICAgICAgICAgICAgY29udGludWUKICAgICAgICAgICAgdyA9ICgzLjAgaWYgdCBpbiBzdHJvbmcgZWxzZSAxLjApICogaWRmW3RdCiAgICAgICAgICAgIG5fdXNlID0gdGV4dC5jb3VudCh0KQogICAgICAgICAgICBoaXRzID0gW2kgKyAxIGZvciBpLCBsbiBpbiBlbnVtZXJhdGUodGV4dC5zcGxpdGxpbmVzKCkpIGlmIGRlZl9yZVt0XS5zZWFyY2gobG4pXQogICAgICAgICAgICBpZiBoaXRzOgogICAgICAgICAgICAgICAgc2NvcmVbZl0gKz0gNiAqIHcgKiBsb3cKICAgICAgICAgICAgICAgIGRlZnNbZl0gKz0gWyh0LCBoKSBmb3IgaCBpbiBoaXRzWzoyXV0KICAgICAgICAgICAgaWYgbl91c2U6CiAgICAgICAgICAgICAgICBzY29yZVtmXSArPSBtaW4obl91c2UsIDMpICogMC4yICogdyAqIGxvdwogICAgICAgICAgICAgICAgdXNlc1tmXSArPSBuX3VzZQogICAgICAgICAgICBpZiBzdGVtX2hpdDoKICAgICAgICAgICAgICAgIHNjb3JlW2ZdICs9IDggKiB3ICogbG93CiAgICByYW5rZWQgPSBzb3J0ZWQoc2NvcmUsIGtleT1sYW1iZGEgZjogLXNjb3JlW2ZdKVs6NV0KICAgIGlmIG5vdCByYW5rZWQ6CiAgICAgICAgcHJpbnQoImxvY2F0ZTogbm8gUHl0aG9uIGZpbGUgbWVudGlvbnMgYW55IG9mIiwgc29ydGVkKHRva3MpWzoxMl0pCiAgICAgICAgcmV0dXJuCiAgICBwcmludCgiQ0FORElEQVRFIEZJTEVTIChiZXN0IGZpcnN0KS4gT3BlbiAjMSB3aXRoIHNlZCAtbiBhcm91bmQgdGhlIERFRklOSVRJT04gbGluZSBhbmQgZWRpdCBUSEVSRToiKQogICAgZm9yIGksIGYgaW4gZW51bWVyYXRlKHJhbmtlZCwgMSk6CiAgICAgICAgZCA9ICIsICIuam9pbihmInt0fSBATHtofSIgZm9yIHQsIGggaW4gZGVmc1tmXVs6NF0pIG9yICJubyBkZWZpbml0aW9uLCB1c2VzIG9ubHkiCiAgICAgICAgcHJpbnQoZiIgIHtpfS4ge2Z9ICAgc2NvcmUge3Njb3JlW2ZdOi4xZn0gfCBkZWZpbmVzOiB7ZH0gfCBtZW50aW9ucyB7dXNlc1tmXX0iKQogICAgcHJpbnQoIkVkaXQgdGhlIGZpbGUgdGhhdCBERUZJTkVTIHRoZSBiZWhhdmlvdXIgKCMxIHVubGVzcyB0aGUgcHJvYmxlbSBuYW1lcyBhbm90aGVyIGZpbGUpLiAiCiAgICAgICAgICAiRG8gbm90IGVkaXQgdGVzdHMsIGRvY3MsIFJFQURNRSwgTElDRU5TRSBvciBwYWNrYWdpbmcgZmlsZXMuIikKCgppZiBfX25hbWVfXyA9PSAiX19tYWluX18iOgogICAgdHJ5OgogICAgICAgIG1haW4oKQogICAgZXhjZXB0IEV4Y2VwdGlvbiBhcyBlOiAgIyBuZXZlciBicmVhayB0aGUgYWdlbnQgbG9vcAogICAgICAgIHByaW50KGYibG9jYXRlOiBlcnJvciB7dHlwZShlKS5fX25hbWVfX306IHtlfSIpCg==")
Path("/kaggle/working/locate.py").write_bytes(LOC)
LANDABLE = set(json.loads("""["fastapi_14077", "fastapi_14301", "fastapi_14786", "rich_3454", "rich_3905", "rich_3006", "rich_3067", "fastapi_9555", "fastapi_9425", "fastapi_9753", "fastapi_14463", "rich_3894", "fastapi_14873", "fastapi_13537", "rich_3882", "rich_2725", "rich_3518", "fastapi_14492", "rich_2943", "rich_4077", "fastapi_14458", "fastapi_14616", "fastapi_11194", "rich_3063", "rich_3506", "rich_3521"]"""))
tasks, tasks_f = [], None
for f in glob.glob("/kaggle/input/**/tasks.jsonl", recursive=True):   # take the copy that carries gold patches
    rows = [json.loads(l) for l in open(f) if l.strip()]
    if sum(bool(r.get("patch")) for r in rows) > len(tasks and [1] or []) and sum(bool(r.get("patch")) for r in rows) > sum(bool(r.get("patch")) for r in tasks):
        tasks, tasks_f = rows, f
snaps = {Path(p).stem.replace(".tar", ""): p for p in glob.glob("/kaggle/input/**/*.tgz", recursive=True)}
print("tasks", len(tasks), "snapshots", len(snaps), "from", tasks_f)

def gold_files(patch):
    fs = []
    for ln in patch.splitlines():
        if ln.startswith("+++ "):
            f = ln[4:].strip()
            f = f[2:] if f.startswith(("a/", "b/")) else f
            b = f.rsplit("/", 1)[-1]
            if f != "/dev/null" and f.endswith(".py") and not (b.startswith("test_") or "/tests/" in "/" + f):
                fs.append(f)
    return fs

def ids_only(q):
    t = set(re.findall(r"`([^`]{2,80})`", q))
    t |= set(re.findall(r"\b[A-Z][a-z0-9]+(?:[A-Z][A-Za-z0-9]+)+\b", q))
    t |= set(re.findall(r"\b[a-z][a-z0-9]*(?:_[a-z0-9]+)+\b", q))
    t |= set(re.findall(r"\b[A-Za-z_][A-Za-z0-9_]*(?=\()", q))
    t |= set(re.findall(r"\b[A-Za-z_]\w*(?:\.[A-Za-z_]\w*)+\b", q))
    return " ".join(sorted(t))

res = []
for t in tasks:
    iid = t["instance_id"]
    gold = gold_files(t.get("patch") or "")
    if iid not in snaps or not gold:
        res.append(dict(id=iid, skip="no snapshot" if iid not in snaps else "no source gold"))
        continue
    d = tempfile.mkdtemp()
    try:
        with tarfile.open(snaps[iid]) as tf:
            tf.extractall(d, filter="data")
        # snapshot may unpack into a single top dir
        subs = [p for p in Path(d).iterdir()]
        root = str(subs[0]) if len(subs) == 1 and subs[0].is_dir() and not (Path(d) / ".git").exists() else d
        row = dict(id=iid, gold=gold, landable=iid in LANDABLE)
        for name, q in (("full", t["problem_statement"]), ("ids", ids_only(t["problem_statement"]))):
            p = subprocess.run([sys.executable, "/kaggle/working/locate.py", "--q", q], cwd=root, capture_output=True,
                               text=True, timeout=120, env=dict(os.environ, PYTHONPATH=root))
            cand = re.findall(r"^\s+\d+\. (\S+)", p.stdout, re.M)
            row[name] = cand
            row[name + "_rank"] = next((i + 1 for i, c in enumerate(cand) if c in gold), None)
        res.append(row)
    except Exception as e:
        res.append(dict(id=iid, skip=f"{type(e).__name__}: {e}"[:120]))
    finally:
        shutil.rmtree(d, ignore_errors=True)

json.dump(res, open("/kaggle/working/locate_offline.json", "w"), indent=1)
ok = [r for r in res if "skip" not in r]
print("evaluated", len(ok), "| skipped", len(res) - len(ok), [r["skip"] for r in res if "skip" in r][:5])
for sub, rs in (("ALL", ok), ("LANDABLE26", [r for r in ok if r["landable"]])):
    for name in ("full", "ids"):
        k1 = sum(r[name + "_rank"] == 1 for r in rs)
        k3 = sum((r[name + "_rank"] or 99) <= 3 for r in rs)
        k5 = sum(r[name + "_rank"] is not None for r in rs)
        print(f"{sub:11s} query={name:4s} n={len(rs):3d}  top1 {k1:3d} ({100*k1/max(1,len(rs)):.0f}%)  top3 {k3:3d} "
              f"({100*k3/max(1,len(rs)):.0f}%)  top5 {k5:3d} ({100*k5/max(1,len(rs)):.0f}%)")
print("\nper task (landable first): id | gold | full rank | ids rank | full top3")
for r in sorted(ok, key=lambda r: (not r["landable"], r["id"])):
    print(f"{'*' if r['landable'] else ' '} {r['id']:16s} {','.join(r['gold'])[:60]:60s} {str(r['full_rank']):>4s} "
          f"{str(r['ids_rank']):>4s}  {r['full'][:3]}")
```
